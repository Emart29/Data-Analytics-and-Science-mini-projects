Welcome to New York City, one of the most-visited cities in the world. There are many Airbnb listings in New York City to meet the high demand for temporary lodging for travelers, which can be anywhere between a few nights to many months. In this project, we will take a closer look at the New York Airbnb market by combining data from multiple file types like `.csv`, `.tsv`, and `.xlsx`.

Recall that **CSV**, **TSV**, and **Excel** files are three common formats for storing data. 
Three files containing data on 2019 Airbnb listings are available to you:

**data/airbnb_price.csv**
This is a CSV file containing data on Airbnb listing prices and locations.
- **`listing_id`**: unique identifier of listing
- **`price`**: nightly listing price in USD
- **`nbhood_full`**: name of borough and neighborhood where listing is located

**data/airbnb_room_type.xlsx**
This is an Excel file containing data on Airbnb listing descriptions and room types.
- **`listing_id`**: unique identifier of listing
- **`description`**: listing description
- **`room_type`**: Airbnb has three types of rooms: shared rooms, private rooms, and entire homes/apartments

**data/airbnb_last_review.tsv**
This is a TSV file containing data on Airbnb host names and review dates.
- **`listing_id`**: unique identifier of listing
- **`host_name`**: name of listing host
- **`last_review`**: date when the listing was last reviewed

```python
# Import necessary packages
import pandas as pd
import numpy as np
```

```python
# Load the CSV file
df_csv = pd.read_csv('data/airbnb_price.csv')

# Load the TSV file
df_tsv = pd.read_csv('data/airbnb_last_review.tsv', sep='\t')

# Load the Excel file
df_excel = pd.read_excel('data/airbnb_room_type.xlsx')

print('Done Loading')
```

```python
# Display the first 5 rows of the df_csv DataFrame
df_csv.head(5)
```

```python
# Display the first 5 rows of the df_tsv DataFrame
df_tsv.head(5)
```

```python
# Display the first 5 rows of the df_excel DataFrame
df_excel.head(5)
```

```python
# observing the datatypes and information
print(df_csv.info())
print(df_tsv.info())
print(df_excel.info())
```

```python
# Merge df_csv and df_tsv DataFrames on the 'listing_id' column using an outer join
new_df = df_csv.merge(df_tsv, on='listing_id', how='outer')
new_df.head()
```

```python
# Merge new_df and df_excel DataFrames on the 'listing_id' column using an outer join
airbnb_df = new_df.merge(df_excel, on  = 'listing_id', how = 'outer')
airbnb_df.head()
```

```python
# Convert the 'last_review' column to datetime format
airbnb_df['last_review'] = pd.to_datetime(airbnb_df['last_review'])

# Display information about the DataFrame to check datatypes and non-null counts
airbnb_df.info()
```

```python
# Find the most recent and earliest review dates in the 'last_review' column
recent_review = airbnb_df['last_review'].max()
earliest_review = airbnb_df['last_review'].min()
```

```python
# Display the most recent and earliest review dates
recent_review, earliest_review
```

```python
airbnb_df['room_type'].value_counts()
```

```python
# standardizing the room type column by changing every value to lower case
airbnb_df['room_type'] = airbnb_df['room_type'].str.lower()
airbnb_df['room_type'].value_counts()
```

```python
# Count the number of listings where the room type is 'private room'
private_room_count = airbnb_df[airbnb_df['room_type'] == 'private room'].shape[0]
private_room_count
```

```python
# Remove the string ' dollars' from the 'price' column and convert the column to float type
airbnb_df['price'] = airbnb_df['price'].str.replace(' dollars', '')
airbnb_df['price'] = airbnb_df['price'].astype('float')
```

```python
airbnb_df.head()
```

```python
airbnb_df.info()
```

```python
# Calculate the average listing price from the 'price' column of airbnb_df
average_listing_price = airbnb_df['price'].mean()
# Display the average price rounded to the nearest integer
f'{average_listing_price:.0f}'
```

```python
# Create a DataFrame summarizing review and listing statistics
review_dates = pd.DataFrame({
    'first_reviewed': [earliest_review],  # Earliest review date
    'last_reviewed': [recent_review],     # Most recent review date
    'nb_private_rooms': [private_room_count],  # Number of private room listings
    'avg_price': [round(average_listing_price, 2)]  # Average listing price, rounded to 2 decimals
})
review_dates
```
